## Specific Conductivity

### NaCl

```python
# loading NaCl equivalent conductivity data
equiv_data = pd.read_csv("./Data/NaCl_equiv.csv")
molar_conc_equiv = equiv_data["Conc"] # mM
equiv_cond = equiv_data["Equiv_cond"] # cm^2.Siemen.equiv^(-1)

# number of data points
ndata = len(equiv_cond)
```

### Validating NaCl Equivalent Conductivity

```python
# constants
temp_NaCl = 298.16 # K
epsilon = 78.3
eta = 8.903*10**(-3) # poise
lambda_0_NaCl = 126.45 # cm^2.Siemen.equiv^(-1)
lambda_0_Na = 50.10 # cm^2.Siemen.equiv^(-1)
lambda_0_Cl = 76.35 # cm^2.Siemen.equiv^(-1)
a_NaCl = 4*10**(-8) # cm
z_Na = 1
z_Cl = -1

# concentration in M
conc = [molar_conc_equiv[i]*10**(-3) for i in range(ndata)]

# equivalent conductivity from equation (4)
def cond_calc_ref(conc, temp, epsilon, eta, lambda_0, a, z_1, z_2, lambda_0_cation, lambda_0_anion):
    """
    Calculates equivalent conductivity using equation (7.36) in reference

    Arguments:
        conc: salt concentration in M
        temp: temperature of salt solution in K
        epsilon: dielectric constant
        eta: viscosity of salt solution in poise
        lambda_0: limiting equivalent conductivity of salt in cm^2.Siemen.equiv^(-1)
        a: distance within which no other ions can penetrate in cm
        z_1: valency of cation
        z_2: valency of anion
        lambda_0_cation: limiting equivalent conductivity of cation in cm^2.Siemen.equiv^(-1)
        lambda_0_anion: limiting equivalent conductivity of anion in cm^2.Siemen.equiv^(-1)
    
    Returns:
        equiv_cond_ref: equivalent conductivity values in cm^2.S/equiv
    """
    # calculating B and q
    B = 50.29*(10**8)*(epsilon*temp)**(-0.5)
    q = np.abs(z_1*z_2)*(lambda_0_cation + lambda_0_anion)/((np.abs(z_1) + np.abs(z_2))*(np.abs(z_2)*lambda_0_cation + np.abs(z_1)*lambda_0_anion))
    
    # B1 and B2
    B_1 = 2.801*10**6*np.abs(z_1*z_2)*q/((epsilon*temp)**(3/2)*(1 + math.sqrt(q)))
    B_2 = 41.25*(np.abs(z_1) + np.abs(z_2))/(eta*(epsilon*temp)**(1/2))
    
    # species concentration
    cation_conc = conc
    Cl_conc = [np.abs(z_1)*conc[i] for i in range(len(conc))]

    # ionic strength
    I = np.zeros(len(conc))
    for i in range(len(conc)):
        conc_all = [cation_conc[i], Cl_conc[i]]
        z = [z_1, z_2]
        I[i] = 1/2*(sum(conc_all[j]*z[j]**2 for j in range(len(conc_all))))

    # equivalent conductivity
    equiv_cond_ref = []
    for i in range(len(conc)):
        equiv_cond_ref.append(lambda_0 - ((B_1*lambda_0 + B_2)*math.sqrt(I[i])/(1 + a*B*math.sqrt(I[i]))))
    
    return(equiv_cond_ref)

# making smoother plots
conc_smooth = np.linspace(0, np.max(conc), 100)
equiv_cond_smooth = cond_calc_ref(conc_smooth, temp_NaCl, epsilon, eta, lambda_0_NaCl, a_NaCl, z_Na, z_Cl, lambda_0_Na, lambda_0_Cl)
```

```python
# making plot comparison of data values and calculated
plt.plot(conc_smooth, equiv_cond_smooth, "r-", label="Python code")
plt.plot(conc, equiv_cond, "b.", markersize=10, label="Theory calculation")
plt.xlabel("NaCl concentration (M)")
plt.ylabel("Equivalent conductivity (cm^2.S/equiv)")
plt.ylim((90, 130))
plt.legend()
plt.show()
```

```python
# loading NaCl specific conductivity data
specific_data = pd.read_csv("./Data/NaCl.csv")
molar_conc_specific_Na = specific_data["NaCl conc"]
specific_cond_exp_Na = specific_data["NaCl cond"]
specific_cond_calibrated_Na = specific_data["NaCl cond cal"]

# number of data points
ndata2 = len(specific_cond_exp_Na)
```

### Calculating NaCl Specific Conductivity

```python
# concentration in M
conc2 = [molar_conc_specific_Na[i]*10**(-3) for i in range(ndata2)]

# function the predicts specific conductivity
def specific_cond_model(conc, temp, epsilon, eta, lambda_0, a, z_1, z_2, lambda_0_cation, lambda_0_anion):
    """
    Calculates specific conductivity from equivalent conductivity

    Arguments:
        conc: salt concentration in M
        temp: temperature of salt solution in K
        epsilon: dielectric constant
        eta: viscosity of salt solution in poise
        lambda_0: limiting equivalent conductivity of salt in cm^2.Siemen.equiv^(-1)
        a: distance within which no other ions can penetrate in cm
        z_1: valency of cation
        z_2: valency of anion
        lambda_0_cation: limiting equivalent conductivity of cation in cm^2.Siemen.equiv^(-1)
        lambda_0_anion: limiting equivalent conductivity of anion in cm^2.Siemen.equiv^(-1)
    
    Returns:
        specific_cond_calc: specific conductivity values in micro Siemen per cm
    """
    
    # equivalent conductivity
    equiv_cond = cond_calc_ref(conc, temp, epsilon, eta, lambda_0, a, z_1, z_2, lambda_0_cation, lambda_0_anion)
    
    # species concentration
    cation_conc = conc
    Cl_conc = [np.abs(z_1)*conc[i] for i in range(len(conc))]
    
    # converting the concentrations in mM to equivalent per litre
    equiv_conc = [cation_conc[i]*np.abs(z_1) for i in range(len(conc))]
    
    # specific conductivity in Siemen per cm
    specific_cond = []
    for i in range(len(conc)):
        specific_cond.append((equiv_conc[i]/1000)*equiv_cond[i])
    
    # specific conductivity in micro Siemen per cm
    specific_cond_calc = [specific_cond[i]*10**6 for i in range(len(conc))]

    return specific_cond_calc

# specific conductivity in micro Siemen per cm
specific_cond_calc_Na = specific_cond_model(conc2, temp_NaCl, epsilon, eta, lambda_0_NaCl, a_NaCl, z_Na, z_Cl, lambda_0_Na, lambda_0_Cl)
```

```python
# making plot comparison of data and calculated values
fig = plt.figure(figsize=(4,4))
plt.plot(molar_conc_specific_Na, specific_cond_calc_Na, "r-", label="Theory")
# plt.plot(molar_conc_specific_Na, specific_cond_exp_Na, "b.", markersize=10, label="Data")
plt.plot(molar_conc_specific_Na, specific_cond_calibrated_Na, "s", markersize=5, label="Data")
plt.xlabel("NaCl Conc. (mM)", fontsize=16,fontweight='bold')
plt.ylabel("Conductivity ($\mathbf{\mu}$S/cm)", fontsize=16,fontweight='bold')
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.tick_params(direction="in",top=True, right=True)
plt.title("Theory Predictions", fontsize=16,fontweight='bold')
plt.legend()
plt.tight_layout()
# plt.savefig("single_model_na1.png")
plt.show()
```

```python
# residuals
residuals = [specific_cond_calc_Na[i] - specific_cond_exp_Na[i] for i in range(ndata2)]

# making plot of residuals of calculated specific conductivity
plt.plot(molar_conc_specific_Na, residuals, "b.")
plt.xlabel("NaCl concentration (mM)")
plt.ylabel("Residuals (uS/cm)")
plt.show()
```

### Evaluating Error

```python
# residual fraction
residuals_percent = [(residuals[i]/specific_cond_exp_Na[i])*100 for i in range(ndata2)]

# visualizing in plot
fig = plt.figure(figsize=(4,4))
plt.plot(molar_conc_specific_Na, residuals_percent, "b.")
plt.xlabel("NaCl concentration (mM)", fontsize=16,fontweight='bold')
plt.ylabel("Residuals/Experiments (%)", fontsize=16,fontweight='bold')
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.tick_params(direction="in",top=True, right=True)
plt.title("Instrument Error", fontsize=16,fontweight='bold')
plt.show()
```

```python
# residuals percent average
residuals_percent.pop(0)
residual_percent_avg = np.average(residuals_percent)
print(residual_percent_avg)

# visualizing in plot
molar_conc_specific_Na.pop(0)
fig = plt.figure(figsize=(4,4))
plt.plot(molar_conc_specific_Na, residuals_percent, "b.")
plt.xlabel("NaCl concentration (mM)", fontsize=16,fontweight='bold')
plt.ylabel("Residuals/Experiments (%)", fontsize=16,fontweight='bold')
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.tick_params(direction="in",top=True, right=True)
plt.title("Instrument Error", fontsize=16,fontweight='bold')
plt.show()
```

```python
# bulk conductivity
specific_cond = [specific_cond_nacl[i] + specific_cond_cacl2[i] for i in range(ndata)]
specific_cond_12 = [specific_cond_nacl_12[i] + specific_cond_cacl2_12[i] for i in range(ndata)]
specific_cond_13 = [specific_cond_nacl_13[i] + specific_cond_cacl2_13[i] for i in range(ndata)]

# making plot comparison of calculated and experimental conductivity of NaCl and CaCl2
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# plotting
ax.plot3D(NaCl_conc, CaCl2_conc, specific_cond, color="red", label="Theory")
ax.plot3D(NaCl_conc12, CaCl2_conc12, specific_cond_12, color="red")
ax.plot3D(NaCl_conc13, CaCl2_conc13, specific_cond_13, color="red")
ax.plot3D(NaCl_conc, CaCl2_conc, cond_exp, "b.", markersize=10, label="Data")
ax.plot3D(NaCl_conc12, CaCl2_conc12, cond_exp12, "b.", markersize=10)
ax.plot3D(NaCl_conc13, CaCl2_conc13, cond_exp13, "b.", markersize=10)
ax.set_xlim(50, 0)
ax.set_ylim(0, 50)
ax.tick_params(axis='x', labelsize=10)
ax.tick_params(axis='y', labelsize=10)
ax.tick_params(axis='z', labelsize=10)
ax.tick_params(direction="in",top=True, right=True)
ax.set_title('3D Conductivity Plot', fontsize=12,fontweight='bold')
ax.set_zlabel("Conductivity ($\mathbf{\mu}$S/cm)", fontsize=12,fontweight='bold')
ax.set_xlabel("NaCl Conc. (mM)", fontsize=12,fontweight='bold')
ax.set_ylabel("CaC$\mathbf{l_2}$ Conc. (mM)", fontsize=12,fontweight='bold')
ax.legend()
plt.tight_layout()
# plt.savefig("single_model_na_ca.png")
plt.show()
```

```python
# residuals for NaCl and CaCl2
e = [specific_cond[i] - cond_exp[i] for i in range(ndata)]
e12 = [specific_cond_12[i] - cond_exp12[i] for i in range(ndata)]
e13 = [specific_cond_13[i] - cond_exp13[i] for i in range(ndata)]

# making plots of residuals
fig = plt.figure()
ax = fig.add_subplot(projection ='3d')

# plotting
ax.plot3D(NaCl_conc, CaCl2_conc, e, "r.")
ax.plot3D(NaCl_conc12, CaCl2_conc12, e12, "b.")
ax.plot3D(NaCl_conc13, CaCl2_conc13, e13, ".", color="black")
ax.set_xlim(50, 0)
ax.set_ylim(0, 50)
ax.set_zlim(0, 600)
ax.tick_params(axis='x', labelsize=10)
ax.tick_params(axis='y', labelsize=10)
ax.tick_params(axis='z', labelsize=10)
ax.tick_params(direction="in",top=True, right=True)
ax.set_title("Residuals Plot", fontsize=12,fontweight='bold')
ax.set_zlabel("Residuals ($\mathbf{\mu}$S/cm)", fontsize=12,fontweight='bold')
ax.set_xlabel("NaCl Conc. (mM)", fontsize=12,fontweight='bold')
ax.set_ylabel("CaC$\mathbf{l_2}$ Conc. (mM)", fontsize=12,fontweight='bold')
plt.tight_layout()
# plt.savefig("residuals_single_model_na_ca.png")
plt.show()
```

```python
# residuals fraction
residuals_percent = [100*e[i]/cond_exp[i] for i in range(ndata)]
residuals_percent12 = [100*e12[i]/cond_exp12[i] for i in range(ndata)]
residuals_percent13 = [100*e13[i]/cond_exp13[i] for i in range(ndata)]

# absolute value of relative error
residuals_percent_abs = [np.abs(residuals_percent[i]) for i in range(ndata)]
residuals_percent2_abs = [np.abs(residuals_percent12[i]) for i in range(ndata)]
residuals_percent3_abs = [np.abs(residuals_percent13[i]) for i in range(ndata)]
print(np.mean(residuals_percent_abs))
print(np.mean(residuals_percent2_abs))
print(np.mean(residuals_percent3_abs))


# making plot
fig = plt.figure()
ax = fig.add_subplot(projection ='3d')

# plotting
ax.plot3D(NaCl_conc, CaCl2_conc, residuals_percent, "r.")
ax.plot3D(NaCl_conc12, CaCl2_conc12, residuals_percent12, "b.")
ax.plot3D(NaCl_conc13, CaCl2_conc13, residuals_percent13, ".", color="black")
ax.set_xlim(50, 0)
ax.set_ylim(0, 50)
ax.tick_params(axis='x', labelsize=10)
ax.tick_params(axis='y', labelsize=10)
ax.tick_params(axis='z', labelsize=10)
ax.tick_params(direction="in",top=True, right=True)
ax.set_title("Relative Error", fontsize=12,fontweight='bold')
ax.set_zlabel("Residuals (%)", fontsize=12,fontweight='bold')
ax.set_xlabel("NaCl Conc. (mM)", fontsize=12,fontweight='bold')
ax.set_ylabel("CaC$\mathbf{l_2}$ Conc. (mM)", fontsize=12,fontweight='bold')
plt.tight_layout()
# plt.savefig("proportion_single_model_na_ca.png")
plt.show()
```
